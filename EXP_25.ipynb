```python
import hashlib
prev = "START"
last_time = 0
n = int(input("Number of records: "))

for i in range(n):
    data = input("Enter evidence record: ")
    t = int(input("Enter timestamp: "))
    h = hashlib.sha256(data.encode()).hexdigest()

    if prev != "START" and (prev != old_hash or t < last_time):
        print("Chain Broken")
        break

    old_hash = h
    prev = h
    last_time = t
else:
    print("Chain Verified")
```
