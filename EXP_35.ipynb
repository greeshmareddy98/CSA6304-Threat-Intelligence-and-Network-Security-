```python
roles = {
    "Admin": ["read", "write", "delete"],
    "Employee": ["read"],
    "Manager": ["read", "write"]
}

users = {
    "Greeshma": ("Employee", ["read", "write"]),
    "Ravi": ("Admin", ["read", "write", "delete"]),
    "Priya": ("Manager", ["read"])
}

for user, (role, permissions) in users.items():
    required = roles[role]

    extra = [p for p in permissions if p not in required]

    print("\nUser:", user)
    print("Role:", role)

    if extra:
        print("Excess Access:", extra)
    else:
        print("Access is Proper")
```
