```python
baseline = {
    "firewall": True,
    "antivirus": True,
    "backup": True
}

current = {
    "firewall": True,
    "antivirus": False,
    "backup": False
}

for control in baseline:
    if current[control] != baseline[control]:
        print(control, "Failed")
```
