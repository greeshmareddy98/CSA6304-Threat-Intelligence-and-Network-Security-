```python
data = [
    "Failed login from unknown IP",
    "Normal user login",
    "Multiple failed login attempts",
    "Suspicious IP detected",
    "Unusual network traffic"
]

print("1. Hypothesis-Based Hunting")

for event in data:
    if "failed login" in event:
        print("Found:", event)

print("\n2. Intelligence-Based Hunting")

for event in data:
    if "suspicious IP" in event or "unknown IP" in event:
        print("Found:", event)

print("\n3. Analytics-Based Hunting")

for event in data:
    if "unusual" in event or "multiple" in event:
        print("Found:", event)
```
