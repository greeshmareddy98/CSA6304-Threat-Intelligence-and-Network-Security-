```python
tools = {
    "nmap": ["network scanning", "port scanning"],
    "wireshark": ["packet analysis", "network traffic"],
    "snort": ["intrusion detection", "ids"],
    "burp suite": ["web security", "web application testing"],
    "john the ripper": ["password cracking", "password testing"]
}

task = input("Enter security task: ").lower()

found = False

for tool, keywords in tools.items():
    for keyword in keywords:
        if keyword in task:
            print("Recommended Security Tool:", tool)
            found = True
            break

if not found:
    print("No suitable tool found")
```
