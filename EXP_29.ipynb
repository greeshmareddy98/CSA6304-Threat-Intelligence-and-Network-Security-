```python
incident = input("Enter incident type: ").lower()

if incident == "malware":
    print("Isolate infected system")
elif incident == "ddos":
    print("Filter traffic and enable protection")
elif incident == "data breach":
    print("Restrict access and secure accounts")
elif incident == "phishing":
    print("Block sender and reset affected passwords")
else:
    print("Investigate and monitor the incident")
```
