```python
logs = input("Enter IP addresses separated by spaces: ").split()
limit = int(input("Enter request limit: "))

for ip in set(logs):
    count = logs.count(ip)
    if count > limit:
        print(ip, "Possible DoS Attack:", count, "requests")
```
