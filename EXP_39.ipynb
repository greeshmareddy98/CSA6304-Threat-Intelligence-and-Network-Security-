```python
feed = [
    "IP: 192.168.1.10",
    "Domain: malware.com",
    "IP: 10.0.0.5"
]

logs = [
    "User connected to 192.168.1.10",
    "Visited google.com",
    "Connection to malware.com",
    "Normal activity"
]

iocs = []

for item in feed:
    value = item.split(": ")[1]
    iocs.append(value.lower())

print("Normalized IOCs:", iocs)

print("\nThreat Intelligence Alerts:")

for log in logs:
    for ioc in iocs:
        if ioc in log.lower():
            print("ALERT! IOC Matched:", ioc)
```
