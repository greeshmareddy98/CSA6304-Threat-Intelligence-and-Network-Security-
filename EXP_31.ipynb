```python
rules = [
    ("DENY", "192.168.1.10", 80),
    ("ALLOW", "192.168.1.10", 80),
    ("ALLOW", "192.168.1.20", 443)
]

ip = input("Enter IP address: ")
port = int(input("Enter port number: "))

result = "DENY"

for action, rule_ip, rule_port in rules:
    if (ip == rule_ip or rule_ip == "*") and (port == rule_port or rule_port == 0):
        result = action
        break

print("Firewall Decision:", result)
```
