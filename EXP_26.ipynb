```python
import hashlib

sample = input("Enter sample text: ")
log = input("Enter behavior log: ")

print("Hash:", hashlib.sha256(sample.encode()).hexdigest())

if any(x in sample.lower() for x in ["eval", "socket", "subprocess"]):
    print("Static: Suspicious API found")
else:
    print("Static: Safe strings")

if any(x in log.lower() for x in ["delete", "connect", "encrypt"]):
    print("Dynamic: Suspicious behavior")
else:
    print("Dynamic: No suspicious behavior")
```
