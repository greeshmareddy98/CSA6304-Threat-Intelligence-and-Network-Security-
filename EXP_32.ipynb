```python
properties = [
    ("Data Encryption", "Confidentiality", True),
    ("File Integrity", "Integrity", False),
    ("Server Backup", "Availability", True),
    ("Access Control", "Confidentiality", False)
]

violations = []

for name, principle, status in properties:
    if status == False:
        violations.append(principle)

print("CIA Principle Violations:")

if violations:
    for v in set(violations):
        print(v, "is violated")
else:
    print("No violations found")
```
