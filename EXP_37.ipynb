```python
baseline = ["login", "logout", "file access"]

logs = [
    "login",
    "logout",
    "unknown access",
    "file access",
    "multiple failed attempts"
]

print("Security Monitoring Started")

for event in logs:
    print("\nEvent:", event)

    if event not in baseline:
        print("ALERT: Anomaly Detected!")
    else:
        print("Normal Activity")

print("\nMonitoring Completed")
```
