```python
description = input("Enter system or data description: ").lower()

if "credit card" in description or "payment" in description:
    standard = "PCI-DSS"
elif "patient" in description or "medical" in description or "health" in description:
    standard = "HIPAA"
elif "personal data" in description or "privacy" in description:
    standard = "GDPR"
elif "financial report" in description or "accounting" in description:
    standard = "SOX"
else:
    standard = "No matching standard found"

print("Applicable Compliance Standard:", standard)
```
