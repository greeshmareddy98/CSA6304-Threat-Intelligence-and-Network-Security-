```python
action = input("Enter incident response action: ").lower()

if "prepare" in action or "training" in action:
    phase = "Preparation"
elif "detect" in action or "identify" in action:
    phase = "Identification"
elif "isolate" in action or "block" in action:
    phase = "Containment"
elif "remove" in action or "delete malware" in action:
    phase = "Eradication"
elif "restore" in action or "recover" in action:
    phase = "Recovery"
elif "review" in action or "document lessons" in action:
    phase = "Lessons Learned"
else:
    phase = "Unknown Phase"

print("Incident Response Phase:", phase)
```
