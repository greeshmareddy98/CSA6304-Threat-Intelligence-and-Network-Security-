```python
likelihood = int(input("Enter likelihood (1-5): "))
impact = int(input("Enter impact (1-5): "))

score = likelihood * impact

print("Risk Score:", score)

if score <= 4:
    level = "Low"
elif score <= 9:
    level = "Medium"
elif score <= 16:
    level = "High"
else:
    level = "Critical"

print("Risk Level:", level)
```
