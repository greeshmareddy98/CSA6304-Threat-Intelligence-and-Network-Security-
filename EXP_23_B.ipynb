```python
trusted = ["192.168.1.10"]
attack = ["malware", "attack"]

ip = input("Enter source IP: ")
traffic = input("Enter traffic: ").lower()

if ip in trusted:
    print("Allowed: Trusted Source")
elif traffic in attack:
    print("Blocked: Attack Detected")
else:
    print("Allowed")
```
